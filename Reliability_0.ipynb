# 2. GUIA RAPIDA PARA EL USUARIO

En este ejemplo, crearemos una distribución Weibull y de esa distribución extraeremos 20 muestras aleatorias. Usando esas muestras ajustaremos una distribución Weibull de 2 parámetros. El proceso de ajuste genera la gráfica de probabilidad. Luego podemos acceder al objeto de distribución para trazar la función de supervivencia.

```python
from reliability.Distributions import Weibull_Distribution
from reliability.Fitters import Fit_Weibull_2P
from reliability.Probability_plotting import plot_points
import matplotlib.pyplot as plt
```

```python
alpha = float(input("Por favor ingrese el valor de alpha: "))
beta = float(input("Por favor ingrese el valor de beta: "))
dist = Weibull_Distribution(alpha, beta)  # creates the distribution object (crea el objeto de distribucion)
data = dist.random_samples(20, seed=42)  # draws 20 samples from the distribution. Seeded for repeatability
plt.subplot(121)
fit = Fit_Weibull_2P(failures=data)  # fits a Weibull distribution to the data and generates the probability plot
plt.subplot(122)
fit.distribution.SF(label='fitted distribution')  # uses the distribution object from Fit_Weibull_2P and plots the survival function
dist.SF(label='original distribution', linestyle='--') # plots the survival function of the original distribution
plot_points(failures=data, func='SF')  # overlays the original data on the survival function
plt.legend()
plt.show()
```

Una característica clave de la confiabilidad es que las distribuciones de probabilidad se crean como objetos, y estos objetos tienen muchas propiedades (como la media) que se establecen una vez que se definen los parámetros de la distribución. El uso del operador de punto nos permite acceder a estas propiedades, así como a una gran cantidad de métodos (como extraer muestras aleatorias como se ve en el ejemplo anterior).

Cada distribución se puede visualizar en cinco gráficos diferentes. Estas son la función de densidad de probabilidad (PDF), la función de distribución acumulativa (CDF), la función de supervivencia (SF) [también conocida como función de confiabilidad], la función de riesgo (HF) y la función de riesgo acumulativo (CHF). Acceder a la trama de cualquiera de estos es tan fácil como cualquiera de los otros métodos. P.ej. dist.SF()en el ejemplo anterior es lo que traza la función de supervivencia utilizando el objeto de distribución que fue devuelto por el instalador.

# 2.1 INTRODUCCION AL CAMPO DE INGENIERIA DE CONFIABILIDAD

La ingeniería de confiabilidad es un campo de estudio que se ocupa de la estimación, prevención y gestión de fallas combinando estadística, análisis de riesgos y física. Al comprender cómo pueden ocurrir o han ocurrido las fallas, podemos predecir mejor la vida útil de un producto o sistema, lo que nos permite administrar su ciclo de vida y los riesgos asociados con su falla. Todos los sistemas, componentes y estructuras de ingeniería eventualmente fallarán, y saber cómo y cuándo ocurrirá esa falla es de gran interés para los propietarios y operadores de esos sistemas. Debido a las similitudes entre el ciclo de vida de los sistemas de ingeniería y el ciclo de vida de los humanos, el campo de estudio conocido como análisis de supervivencia tiene muchos conceptos que se utilizan en la ingeniería de confiabilidad.

Todo el mundo es muy consciente de la importancia de la confiabilidad, especialmente cuando algo no funciona en el momento que esperamos. Ya sea que su automóvil no arranque, su televisor falle o la posibilidad de sufrir un retraso en la pista porque la unidad de aire acondicionado de su avión simplemente dejó de funcionar, sabemos que la falla del sistema es algo que todos queremos evitar. Cuando no se puede evitar, al menos queremos saber cuándo es probable que ocurra para poder realizar un mantenimiento preventivo antes de que surja la necesidad de un mantenimiento correctivo. La ingeniería de confiabilidad se utiliza con mayor frecuencia para sistemas que son de importancia crítica para la seguridad (como en la industria nuclear), o en sistemas que son numerosos (como vehículos o electrónica) donde el costo de los problemas de confiabilidad en toda la flota puede volverse muy costoso rápidamente.

Gran parte de la ingeniería de confiabilidad implica el análisis de datos (como los datos del tiempo hasta la falla) para descubrir los patrones de cómo ocurren las fallas. Una vez que entendemos cómo fallan las cosas, podemos usar esos patrones para pronosticar cómo ocurrirán las fallas a lo largo de la vida útil de una población de artículos, o la vida útil de uno o más artículos reparables. Esta biblioteca de Python está diseñada para ayudar con la parte del análisis de datos de la ingeniería de confiabilidad.

Más información está disponible en Wikipedia y en muchos otros recursos de confiabilidad .

# 2.3.9 Ubicación cambiando las distribuciones

Dentro de la confiabilidad, la parametrización de las distribuciones Exponencial, Weibull, Gamma, Lognormal y Loglogistic permite el cambio de ubicación utilizando el parámetro gamma. Esto simplemente desplazará el límite inferior de la distribución hacia la derecha de 0 a . En la forma de las distribuciones con desplazamiento de ubicación, las ecuaciones enumeradas anteriormente son casi idénticas, excepto en todos los lugares donde se ve (t) reemplazarlo con (t-gamma) . La razón para utilizar la forma de distribución desplazada de ubicación es porque algunos fenómenos que pueden modelarse bien mediante una determinada distribución de probabilidad no comienzan a ocurrir inmediatamente, por lo que se hace necesario desplazar el límite inferior de la distribución para que los datos puedan ser modelado con precisión por la distribución.

Si implementa esto usted mismo, asegúrese de establecer todos los valores de (y) en 0 para (t<=gamma) ya que la forma cruda de las distribuciones desplazadas de ubicación anteriores no pondrá a cero automáticamente estos valores y puede resultar en valores negativos. Esta puesta a cero se realiza automáticamente dentro de la confiabilidad .

En este ejemplo, los valores y devueltos se suman para producir la matriz 'combinada' que luego se traza usando matplotlib contra los xvals. Dado que el 89% de los componentes analizados no muestran evidencia de desgaste, es sorprendente cuántos de nuestros programas de mantenimiento hasta el día de hoy todavía tienen políticas de mantenimiento basadas en el desgaste y cuán omnipresente se ha vuelto el término "curva de bañera" en la comunidad de mantenimiento. Todas las distribuciones se pueden ajustar tanto a datos completos como a datos incompletos (censurados a la derecha). Todas las distribuciones en el módulo Fitters se nombran con su número de parámetros (por ejemplo, Fit_Weibull_2P usa α,β, mientras que Fit_Weibull_3P usa α,β,γ). Esto tiene como objetivo eliminar la ambigüedad sobre qué distribución está adaptando.

Las distribuciones se ajustan simplemente usando la función deseada y especificando los datos como fallas o datos censurados por la derecha. Debe tener al menos tantas fallas como parámetros de distribución, o el ajuste estaría insuficientemente restringido. Generalmente es recomendable tener al menos 4 puntos de datos ya que la precisión del ajuste es proporcional a la cantidad de datos. Una vez instalados, los resultados se asignan a un objeto y se puede acceder a los parámetros ajustados por nombre, como se muestra en los ejemplos siguientes. Los criterios de bondad de ajuste también están disponibles como AICc (Criterio de información de Akaike corregido), BIC (Criterio de información bayesiano), AD (Anderson-Darling) y loglik (logaritmo de verosimilitud), aunque son más útiles al comparar el ajuste de múltiples distribuciones como en la Fit_Everything . Por conveniencia, cada uno de los módulos en Fitters también genera un objeto de distribución que tiene los parámetros de la distribución ajustada.

La distribución Beta es solo para datos en el rango de 0 a 1. Especificar datos fuera de este rango provocará un error. La distribución Beta ajustada no incluye intervalos de confianza.

Si tiene una cantidad muy grande de datos (>100.000 muestras), es probable que su computadora tarde mucho tiempo en calcular los resultados. Esta es una limitación de los lenguajes interpretados como Python en comparación con los lenguajes compilados como C++, en los que están escritos muchos paquetes de software de confiabilidad comercial. Si tiene volúmenes de datos muy grandes, es posible que desee considerar el uso de software comercial para un tiempo de cálculo más rápido. La función Fit_Weibull_2P_grouped está diseñada para aceptar un marco de datos que tiene múltiples apariciones de algunos valores (por ejemplo, múltiples valores censurados al mismo valor cuando finalizó la prueba). Dependiendo del tamaño del conjunto de datos y la cantidad de agrupación de sus datos, Fit_Weibull_2P_grouped puede ser mucho más rápido que Fit_Weibull_2P y alcanzar el mismo nivel de precisión. Esta diferencia no se nota si tiene menos de 10000 muestras. Para obtener más información, consulte el siguiente ejemplo sobre el uso de Fit_Weibull_2P_grouped .

Los datos muy censurados (>99,9% de censura) pueden provocar que el optimizador no encuentre una solución. Si tiene datos muy censurados, es posible que tenga un problema limitado de población fallida. Se recomienda que no intente ajustar una de estas distribuciones estándar a dicho conjunto de datos, ya que sus resultados (aunque hayan logrado un ajuste exitoso) serán una mala descripción de su estadística de población general y corre el riesgo de sacar conclusiones equivocadas cuando Se ha instalado un modelo equivocado. Si no sabe qué distribución desea ajustar, consulte la sección sobre el uso de la función Fit_Everything, que encontrará la mejor distribución para describir sus datos. Es muy recomendable que siempre intentes adaptar todo y aceptar el mejor ajuste en lugar de elegir una distribución concreta por motivos subjetivos.

Si ha intentado ajustar múltiples distribuciones y nada parece funcionar bien, o puede ver que los puntos de dispersión en el gráfico de probabilidad tienen forma de S o una curva, entonces es posible que tenga datos de una combinación de fuentes. Ejemplo 4 

Como otro ejemplo, ajustaremos una distribución Gamma_2P a algunos datos parcialmente censurados por la derecha. Para proporcionar una comparación de la precisión del ajuste a medida que aumenta el número de muestras, haremos el mismo experimento con diferentes tamaños de muestra. Los resultados destacan que la precisión del ajuste es proporcional a la cantidad de muestras, por lo que siempre se debe intentar obtener más datos si es posible.

```python
from reliability.Distributions import Gamma_Distribution
from reliability.Fitters import Fit_Gamma_2P
from reliability.Other_functions import make_right_censored_data, histogram
import matplotlib.pyplot as plt
```

```python
a = 30
b = 4
threshold = 180  # this is used when right censoring the data
trials = [10, 100, 1000, 10000]
subplot_id = 221
plt.figure(figsize=(9, 7))
for sample_size in trials:
    dist = Gamma_Distribution(alpha=a, beta=b)
    raw_data = dist.random_samples(sample_size, seed=2)  # create some data. Seeded for repeatability
    data = make_right_censored_data(raw_data, threshold=threshold)  # right censor the data
    gf = Fit_Gamma_2P(failures=data.failures, right_censored=data.right_censored, show_probability_plot=False, print_results=False)  # fit the Gamma_2P distribution
    print('\nFit_Gamma_2P parameters using', sample_size, 'samples:', '\nAlpha:', gf.alpha, '\nBeta:', gf.beta)  # print the results
    plt.subplot(subplot_id)
    histogram(raw_data,white_above=threshold) # plots the histogram using optimal bin width and shades the right censored part white
    dist.PDF(label='True')  # plots the true distribution
    gf.distribution.PDF(label='Fitted', linestyle='--')  # plots the fitted Gamma_2P distribution
    plt.title(str(str(sample_size) + ' samples\n' + r'$\alpha$ error: ' + str(round(abs(gf.alpha - a) / a * 100, 2)) + '%\n' + r'$\beta$ error: ' + str(round(abs(gf.beta - b) / b * 100, 2)) + '%'))
    plt.ylim([0, 0.012])
    plt.xlim([0, 500])
    plt.legend()
    subplot_id += 1
plt.subplots_adjust(left=0.11, bottom=0.08, right=0.95, top=0.89, wspace=0.33, hspace=0.58)
plt.show()
```

# 7. Modelos Mixtos

Los modelos mixtos son una combinación de dos o más distribuciones sumadas para crear una distribución que tiene una forma con más flexibilidad que una distribución única. Cada uno de los componentes de la mezcla debe multiplicarse por una proporción, y la suma de todas las proporciones es igual a 1. La mezcla generalmente se escribe en términos de la PDF, pero como la CDF es la integral (suma acumulativa) de la PDF, De manera equivalente, podemos escribir el modelo Mixture en términos de PDF o CDF. Para un modelo de mezcla con 2 distribuciones, las ecuaciones se muestran a continuación:

Imagen1.png

Los modelos mixtos son útiles cuando hay más de un modo de falla que genera los datos de falla. Esto se puede reconocer porque la forma del PDF y el CDF está fuera de lo que cualquier distribución única puede modelar con precisión. En un gráfico de probabilidad, se puede identificar una combinación de modos de falla mediante curvaturas o formas de S en los datos que de otro modo se esperaría que fueran lineales. Un ejemplo de esto se muestra en la imagen a continuación. No debe utilizar un modelo mixto sólo porque puede adaptarse a casi cualquier cosa realmente bien, pero debe utilizar un modelo mixto si sospecha que existen múltiples modos de falla que contribuyen a los datos de falla que está observando. Para juzgar si un modelo mixto está justificado, observe el criterio de bondad de ajuste (AICc o BIC) que penaliza la puntuación en función del número de parámetros del modelo. Cuanto más cercano a cero sea el criterio de bondad de ajuste, mejor será el ajuste. Usar AD o log-verosimilitud para esta verificación no es apropiado ya que estos criterios de bondad de ajuste no penalizan la puntuación en función del número de parámetros en el modelo y, por lo tanto, son propensos a un sobreajuste.

Véase también los modelos de riesgo competitivos para conocer otro método de combinar distribuciones utilizando el producto del SF en lugar de la suma del CDF.

# 7.1 Creando un modelo de mezcla 

Dentro de confiabilidad.Distribuciones está el Mixture_Model. Esta función acepta una matriz o lista de objetos de distribución estándar creados utilizando el de confiabilidad. Distribuciones (las distribuciones disponibles son Exponencial, Weibull, Gumbel, Normal, Lognormal, Loglogistic, Gamma, Beta). No hay límite para la cantidad de componentes que puede agregar a la mezcla, pero generalmente es preferible usar tan pocos como sean necesarios para ajustarse adecuadamente a los datos (generalmente 2 o 3). Además de las distribuciones, puedes especificar las proporciones que aporta cada distribución en la mezcla. Estas proporciones deben sumar 1. Si no se especifica, las proporciones se establecerán como iguales para cada componente.

Como este proceso es aditivo para la función de supervivencia y puede aceptar muchas distribuciones de diferentes tipos, la formulación matemática rápidamente se vuelve compleja. Por esta razón, el algoritmo combina los modelos numéricamente en lugar de empíricamente, por lo que no existen fórmulas simples para muchas de las estadísticas descriptivas (media, mediana, etc.). Además, la precisión del modelo depende de xvals. Si la matriz xvals es pequeña (<100 valores), la respuesta será "en bloques" e inexacta. La variable xvals sólo se acepta para PDF, CDF, SF, HF y CHF. Los otros métodos (como las muestras aleatorias) utilizan los valores x predeterminados para una máxima precisión. El número predeterminado de valores generados cuando no se proporciona xvals es 1000. Considere esto cuidadosamente al especificar xvals para evitar imprecisiones en los resultados.

Ejemplo 1 

El siguiente ejemplo muestra cómo se puede crear, visualizar y utilizar el objeto Mixture_Model.

```python
from reliability.Distributions import Lognormal_Distribution, Gamma_Distribution, Weibull_Distribution, Mixture_Model
import matplotlib.pyplot as plt

# create the mixture model
d1 = Lognormal_Distribution(mu=2, sigma=0.8)
d2 = Weibull_Distribution(alpha=50, beta=5, gamma=100)
d3 = Gamma_Distribution(alpha=5, beta=3, gamma=30)
mixture_model = Mixture_Model(distributions=[d1, d2, d3], proportions=[0.3, 0.4, 0.3])

# plot the 5 functions using the plot() function
mixture_model.plot()

# plot the PDF and CDF
plot_components = True # this plots the component distributions. Default is False
plt.figure(figsize=(9, 5))
plt.subplot(121)
mixture_model.PDF(plot_components=plot_components, color='red', linestyle='--')
plt.subplot(122)
mixture_model.CDF(plot_components=plot_components, color='red', linestyle='--')
plt.subplots_adjust(left=0.1, right=0.95)
plt.show()

# extract the mean of the distribution
print('The mean of the distribution is:', mixture_model.mean)

'''
The mean of the distribution is: 74.91607709895453
'''
```

La fórmula para el modelo de riesgos competitivos generalmente se escribe en términos de la función de supervivencia (FS). # Creación de un modelo de riesgos competitivos

Dentro de las Distribuciones de confiabilidad se encuentra el Modelo_de_riesgos_competitivos. Esta función acepta una matriz o lista de objetos de distribución creados utilizando el módulo de confiabilidad. A diferencia del modelo de mezcla, no es necesario especificar ninguna proporción.

Como este proceso es multiplicativo para la función de supervivencia (o aditivo para la función de riesgo) y puede aceptar muchas distribuciones de diferentes tipos, la formulación matemática rápidamente se vuelve compleja. Por esta razón, el algoritmo combina los modelos numéricamente en lugar de empíricamente, por lo que no existen fórmulas simples para muchas de las estadísticas descriptivas (media, mediana, etc.). Además, la precisión del modelo depende de xvals. Si la matriz xvals es pequeña (<100 valores), la respuesta será "en bloques" e inexacta. La variable xvals sólo se acepta para PDF, CDF, SF, HF y CHF. Los otros métodos (como las muestras aleatorias) utilizan los valores x predeterminados para una máxima precisión. El número predeterminado de valores generados cuando no se proporciona xvals es 1000. Considere esto cuidadosamente al especificar xvals para evitar imprecisiones en los resultados.

# Modelos DSZI

¿Qué son los modelos DSZI? 

DSZI es un acrónimo de “Subpoblación defectuosa inflada cero”. Es una combinación del modelo de subpoblación defectuosa (DS) y el modelo de inflación cero (ZI).

Un modelo de subpoblación defectuoso es aquel en el que la CDF no llega a 1 durante el período de observación. Esto se produce cuando una parte de la población fracasa (conocida como subpoblación defectuosa), pero el resto de la población no fracasa (y es censurada correctamente) al final del período de observación.

Un modelo inflado a cero es aquel en el que el CDF comienza por encima de 0 al inicio del período de observación. Esto se debe a que muchos elementos de la población “muertos al llegar” están representados por tiempos de falla de 0. Esto no es lo mismo que los datos censurados a la izquierda, ya que los datos censurados a la izquierda son cuando las fallas ocurrieron entre 0 y el tiempo de observación. En el modelo inflado a cero, se considera que el tiempo de observación comienza en 0, por lo que los tiempos de falla son 0.

En un modelo DSZI, el CDF (que normalmente va de 0 a 1) va de encima de 0 a debajo de 1, como se muestra en la imagen siguiente. En esta imagen, la escala del PDF y del CDF está normalizada para que ambos puedan verse juntos. En realidad, el CDF es mucho más grande que el PDF.

imagen.png

Se puede aplicar un modelo DSZI a cualquier distribución (Weibull, Normal, Lognormal, etc.) utilizando las transformaciones que se explican en la siguiente sección. El siguiente gráfico muestra cómo una distribución Weibull puede convertirse en DS_Weibull, ZI_Weibull y DSZI_Weibull. Tenga en cuenta que la PDF de los modelos DS, ZI y DSZI parece más pequeña que la del modelo Weibull original, ya que el área debajo de la PDF ya no es 1. Esto se debe a que el CDF no oscila entre 0 y 1.

```python
import matplotlib.pyplot as plt
from reliability.Distributions import Weibull_Distribution
from reliability.Other_functions import make_right_censored_data
from reliability.Fitters import Fit_Everything
from reliability.Repairable_systems import optimal_replacement_time

# 1. Ingesta y preparacion de datos observados con censura por umbral
dist_teorica = Weibull_Distribution(alpha=120, beta=2.2)
muestras_campo = dist_teorica.random_samples(100, seed=42)
datos_preparados = make_right_censored_data(muestras_campo, threshold=100)

# 2. Evaluacion automatica de bondad de ajuste de distribuciones
modelos_evaluados = Fit_Everything(
    failures=datos_preparados.failures,
    right_censored=datos_preparados.right_censored,
    show_probability_plot=False
)

# 3. Seleccion automatica del modelo optimo (Menor AICc)
distribucion_optima = modelos_evaluados.best_distribution
print("Modelo seleccionado:", modelos_evaluados.best_distribution_name)

# 4. Calculo de estrategia optima de reemplazo preventivo por minimizacion de costos
# Costo de Mantenimiento Preventivo = $250, Costo de Mantenimiento Correctivo = $1500
intervalo_optimo = optimal_replacement_time(
    cost_PM=250,
    cost_CM=1500,
    Weibull_2P=distribucion_optima
)

plt.show()
```
