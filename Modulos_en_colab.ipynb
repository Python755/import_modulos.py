**1. Cómo crear y manipular archivos .py en Colab**
No necesitas salir de Colab para crear tus scripts. Tienes dos formas principales:
A. Usando el Editor de Archivos (Interfaz Visual)
Haz clic en el icono de Carpeta en la barra lateral izquierda.
Haz clic derecho en el panel de archivos y selecciona Nuevo archivo.
Ponle nombre, por ejemplo: mis_funciones.py.
Haz doble clic en el archivo creado. Se abrirá un editor de texto a la derecha donde puedes escribir tu código.
Presiona Ctrl + S para guardar (aunque Colab suele autoguardar al cerrar el panel).


**B. Usando comandos mágicos (Desde una celda)**
Puedes escribir el código directamente en una celda de código y guardarlo automáticamente en un archivo usando %%writefile:

```python
%%writefile operaciones.py

def saludar(nombre):
    return f"Hola {nombre}, bienvenido a mi módulo."

def sumar(a, b):
    return a + b
saludar('PPI, Python para ingenieros')
# Al ejecutar esta celda, se creará (o sobrescribirá)
# el archivo operaciones.py en el directorio actual.
```

**2. Cómo importar tu módulo en el Notebook**
Una vez creado el archivo, puedes importarlo como cualquier librería de Python:

```python
import operaciones

print(operaciones.saludar("PPI, Python para ingenieros"))
print("Resultado:", operaciones.sumar(10, 5))
```

**3. El problema del "Caché" (Y cómo solucionarlo)**
Importante: Python carga los módulos en memoria la primera vez que los importas. Si haces cambios en tu archivo .py y vuelves a ejecutar import operaciones, los cambios no se verán reflejados.

Para solucionar esto sin reiniciar el entorno, usa la extensión autoreload:

```python
import importlib
import operaciones

# Debido a que la extensión `autoreload` no es compatible con Python 3.12
# (ya que el módulo 'imp' ha sido removido),
# puedes recargar el módulo manualmente después de hacer cambios en 'operaciones.py'.
# Por ejemplo, después de modificar y guardar 'operaciones.py', ejecuta la siguiente línea:
# importlib.reload(operaciones)

# Luego, puedes volver a llamar a las funciones para ver los cambios.
# print(operaciones.saludar("Gemini"))
```

# Ejemplo practico

**Paso 1: Crear el módulo inicial**
En una celda de código, escribe lo siguiente y ejecútala. Esto creará un archivo físico en el servidor de Colab llamado conversor.py.

```python
%%writefile conversor.py

def celsius_a_fahrenheit(celsius):
    """Convierte grados Celsius a Fahrenheit."""
    return (celsius * 9/5) + 32
```

**Paso 2: Configurar la "Recarga Automática"**
Este es el paso más importante. Si no haces esto, cuando modifiques el archivo más adelante, Colab no se enterará y seguirá usando la versión vieja.

Ejecuta esto en una nueva celda:

```python
# %load_ext autoreload
# %autoreload 2

# import conversor

# Probamos el módulo
# resultado = conversor.celsius_a_fahrenheit(0)
# print(f"0°C son {resultado}°F") # Debería imprimir 32.0
```

```python
import conversor
importlib.reload(conversor)
# probamos el modulo
resultado = conversor.celsius_a_fahrenheit(0)
print(f"0°C son {resultado}°F")
# Deberia imprimir 32.0
```

**Paso 3: Modificar el módulo**
Ahora imagina que quieres agregar una nueva función a ese mismo archivo (por ejemplo, para convertir a Kelvin). Tienes dos formas de hacerlo:

**Opción A: Usando el editor visual (Recomendado para cambios rápidos)**
Haz clic en el icono de la Carpeta (izquierda).

Haz doble clic en conversor.py. Se abrirá un panel a la derecha.

Escribe la nueva función directamente ahí:

```python
# la carpeta donde estan los archivos de la sesion de trabajo es content
# Escribi en el panel de la derecha lo siguiente:
def celsius_a_kelvin(celsius):
    return celsius + 273.15
```

Presiona Ctrl + S (o Cmd + S) para guardar los cambios en el panel del editor.

**Opción B: Sobrescribiendo con código**
Simplemente vuelve a ejecutar la celda del Paso 1 pero agregando la nueva función:

```python
%%writefile conversor.py

def celsius_a_fahrenheit(celsius):
    return (celsius * 9/5) + 32

def celsius_a_kelvin(celsius):
    return celsius + 273.15
```

**Paso 4: Probar la modificación**
Gracias a que activamos %autoreload 2 en el paso 2, no necesitas reiniciar nada. Solo usa la nueva función:

```python
import conversor # No hace daño volverlo a poner
importlib.reload(conversor)
print(f"100°C en Kelvin son: {conversor.celsius_a_kelvin(100)}")
```

Para guardar los archivos que creas en la sesión de Colab (como tus módulos .py) de forma permanente en tu Google Drive, debes seguir estos pasos. Recuerda que si no lo haces, los archivos se borrarán en cuanto cierres la pestaña o pasen unos minutos de inactividad.

Primero, debes darle permiso a Colab para acceder a tu espacio de Drive. Ejecuta este código:

```python
from google.colab import drive
drive.mount('/content/drive')
```

**2. Copiar los archivos de la sesión a Drive**
Una vez montado, tu Drive aparecerá como una carpeta llamada /content/drive/MyDrive.

Si tienes un archivo llamado geometria.py en la sesión actual y quieres guardarlo en una carpeta de tu Drive llamada "MiProyecto", usa el comando cp (copy):

```python
# Crear la carpeta en Drive (si no existe)
!mkdir -p "/content/drive/MyDrive/MiProyecto"

# Copiar el archivo desde la sesión a Drive
!cp geometria.py "/content/drive/MyDrive/MiProyecto/geometria.py"
```

**3. Crear archivos directamente en Drive (El método más inteligente)**
En lugar de crearlos en la sesión temporal y luego copiarlos, puedes indicarle a %%writefile que guarde el archivo directamente en tu Drive:

```python
%%writefile /content/drive/MyDrive/MiProyecto/geometria.py
import math

def area_circulo(radio):
    return math.pi * radio**2
```

**4. Cómo usar ese módulo guardado en el futuro**
Cuando abras un cuaderno nuevo otro día, para poder usar ese archivo .py que guardaste en Drive, deberás hacer dos cosas:

Montar el Drive (el código del paso 1).

Añadir la carpeta al sistema para que Python sepa dónde buscar:

```python
import sys
sys.path.append('/content/drive/MyDrive/MiProyecto')

import geometria
print(geometria.area_circulo(10))
```
