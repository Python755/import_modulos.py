# Yt-dpl Convencional

```python
# instaladndo libreria en entorno de Google colab
!pip install -U "yt-dlp[default]"
```

```python
import yt_dlp

# Install Node.js to provide a JavaScript runtime for yt-dlp
!apt-get install -y nodejs

# 1. Configura el enlace del video o canción
url_audio = input("Por favor pegue aqui el link del video de youtube, que desea descargar y convertir a audio mp3: ")
# 2. Configuración específica para extraer solo el audio en MP3
ydl_opts = {
    'format': 'bestaudio/best',  # Descarga solo el flujo de mejor audio disponible
    'outtmpl': '%(title)s.%(ext)s', # Guarda el archivo con el título original
    'postprocessors': [{         # Este bloque le ordena a Python convertir el archivo a MP3
        'key': 'FFmpegExtractAudio',
        'preferredcodec': 'mp3',
        'preferredquality': '320', # Máxima calidad de audio (320 kbps)
    }],
    'js_runtimes': {'node': {}}, # Specify Node.js as the JavaScript runtime with an empty config dict
}

# 3. Ejecuta el proceso
print("Extrayendo y convirtiendo el audio a MP3 en los servidores de Google...")
with yt_dlp.YoutubeDL(ydl_opts) as ydl:
    ydl.download([url_audio])
print("¡Conversión a MP3 completada con éxito!")
```

# Opciones para corregir el error

No existe una solución "permanente" e infalible debido a que YouTube actualiza continuamente sus algoritmos anti-bot, la detección de firmas de código y los bloqueos a los rangos de direcciones IP públicas de centros de datos como los de Google Colab.

Sin embargo, puedes aplicar variantes en el código de yt-dlp en Python que resuelven o minimizan estos errores en casi la totalidad de los casos:

**1. Cambiar el cliente de reproducción (Player Client Spoofing)**
La causa más frecuente del error en Google Colab es que la interfaz web por defecto de YouTube detecta que la petición proviene de un servidor en la nube. Cambiar el cliente a android, ios o tv suele saltarse las restricciones sin necesidad de cookies.\

```python
# 1. Asegurar la última versión de yt-dlp y ffmpeg
!pip install -U yt-dlp
!apt-get update -qq && apt-get install -y ffmpeg > /dev/null

import yt_dlp

url = input("Ingresa la URL del video de YouTube: ").strip()

ydl_opts = {
    'format': 'bestaudio/best',
    # IMPORTANTE: No incluir 'cookiefile'
    'extractor_args': {
        'youtube': {
            'player_client': ['mweb', 'tv_embedded', 'tv']
        }
    },
    'postprocessors': [{
        'key': 'FFmpegExtractAudio',
        'preferredcodec': 'mp3',
        'preferredquality': '320',
    }],
    'outtmpl': '%(title)s.%(ext)s',
}

try:
    with yt_dlp.YoutubeDL(ydl_opts) as ydl:
        ydl.download([url])
    print("\n✅ ¡Descarga completada!")
except Exception as e:
    print(f"\n❌ Error: {e}")
```

```python
# 1. Es fundamental actualizar yt-dlp e instalar ffmpeg en la máquina de Colab
!pip install -U yt-dlp
!apt-get update -qq && apt-get install -y ffmpeg > /dev/null

import yt_dlp

url = input("Ingresa la URL del video de YouTube: ").strip()

ydl_opts = {
    'format': 'bestaudio/best',
    # Emular cliente iOS/Android evita la detección de bot en IPs de servidores
    'extractor_args': {
        'youtube': {
            'player_client': ['ios', 'android', 'mweb']
        }
    },
    'postprocessors': [{
        'key': 'FFmpegExtractAudio',
        'preferredcodec': 'mp3',
        'preferredquality': '320',
    }],
    'outtmpl': '%(title)s.%(ext)s',
    'quiet': False,
}

try:
    with yt_dlp.YoutubeDL(ydl_opts) as ydl:
        ydl.download([url])
    print("\n✅ ¡Descarga y conversión a MP3 completadas!")
except Exception as e:
    print(f"\n❌ Error al descargar: {e}")
```

2. Usar un Proxy o VPN dentro de Google Colab
Si la IP de Google Colab está bloqueada, puedes enrutar el tráfico de yt-dlp a través de un proxy para cambiar la IP de salida.

Usando un proxy en Python:

```python
import yt_dlp

ydl_opts = {
    'format': 'bestaudio/best',
    'postprocessors': [{
        'key': 'FFmpegExtractAudio',
        'preferredcodec': 'mp3',
        'preferredquality': '320',
    }],
    # Sustituye con un proxy HTTP/SOCKS5 funcional
    'proxy': 'http://usuario:contraseña@ip_proxy:puerto',
    'outtmpl': '%(title)s.%(ext)s',
}

url = input("Por favor pega aqui el link del video: ")

with yt_dlp.YoutubeDL(ydl_opts) as ydl:
    ydl.download([url])
```

# Reproducir DVD desde Python

```python
!pip install python-vlc
```

```python
import time
import vlc

# 1. Inicializar la instancia de VLC
instancia = vlc.Instance("--no-video-title-show")
reproductor = instancia.media_player_new()

# 2. Definir la ruta de tu unidad de DVD (cambia 'D' por la letra de tu lectora)
# El prefijo 'dvd:///' le indica a VLC que lea el disco físico
ruta_dvd = "dvd:///D:/"

# 3. Cargar el DVD y reproducir
media = instancia.media_new(ruta_dvd)
reproductor.set_media(media)
reproductor.play()

# 4. Mantener el script activo mientras se reproduce
print("Reproduciendo DVD... Presiona Ctrl+C para salir.")
try:
    while True:
        time.sleep(1)
except KeyboardInterrupt:
    reproductor.stop()
    print("\nReproducción detenida.")
```
